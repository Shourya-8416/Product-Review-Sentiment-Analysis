```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import re
import pickle
```

```python
nltk.download('stopwords')

# Load dataset
data = pd.read_csv('data/reviews.csv')
data.dropna(inplace=True)
```

```python
# Preprocessing
def preprocess(text):
    text = re.sub(r'[^a-zA-Z]', ' ', text)
    text = text.lower()
    tokens = text.split()
    ps = PorterStemmer()
    tokens = [ps.stem(word) for word in tokens if word not in stopwords.words('english')]
    return ' '.join(tokens)

data['cleaned_review'] = data['review'].apply(preprocess)
```

```python

# TF-IDF
vectorizer = TfidfVectorizer(max_features=3000)
X = vectorizer.fit_transform(data['cleaned_review']).toarray()
y = data['sentiment'].map({'positive': 1, 'negative': 0})
```

```python
# Model training
model = LogisticRegression()
model.fit(X, y)
```

```python
# Evaluation
y_pred = model.predict(X)
print(classification_report(y, y_pred))
sns.heatmap(confusion_matrix(y, y_pred), annot=True)
plt.show()
```

```python
# Save model
pickle.dump(model, open('models/sentiment_model.pkl', 'wb'))
pickle.dump(vectorizer, open('models/vectorizer.pkl', 'wb'))
```
